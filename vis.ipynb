Netflix Data Visulization Project

INTRO : this project analyzes Netflix shows dataset using data visulization techniques. Different charts and graphs 
        are used to identify different trends over all these years 

OBJECTIVE : the motive of this project is to transfrom raw dataset into meaningful visual insights using libraries
            like pandas and matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("n_movies.csv")
df = df.head(1000)
print(df.shape)

print(df.columns)
```

WHAT I DID -
I imported a csv file named as n_movies.csv conataining large dataset of netflix series and movies.

```python
top_ten = df['genre'].value_counts().head(10)
plt.pie(top_ten, labels=top_ten.index, autopct='%1.1f%%') #decimal place
plt.title("Top 10 Genre in Netflix")
plt.show()
```

Observations Made -
Genre Distribution in Netflix shown using a  pie chart
We observed that Crime,Drama and Mystery genre dominates the genre list with 14.9% viewers accross the world
Comedy, Romance and Crime, Drama, Thriller genre is least popular with just 7.0% viewers

```python
df['year'].value_counts().sort_index().plot()
plt.title("Content over the last years")
plt.xlabel("Year")
plt.ylabel("Titles")
plt.xticks(rotation=45)
plt.show()
```

Observations Made -
we can observe from the line graph that netflix came into existence during 1997
its content grew rapidly after 2008
it peaked with its shows during 2022 with 600+ shows

```python
r = df.sort_values(by='rating', ascending=False).head(10)
plt.figure(figsize=(12,5))
plt.bar(r['title'],  r['rating'], color='purple')
plt.title("Top 10 Highest Rated Netflix Shows")
plt.xlabel("Show Name")
plt.ylabel("Rating")
plt.xticks(rotation=90)
plt.show()
```

Observations made -
here is the top 10 highest rated shows available in netflix, displayed using bar graph
here a shoe named 1899 topped this list with 9+ ratings
other mentioned shows also have great ratings
